🔧 Data Cleaning: Handling Missing Data

```python
movies.shape
```

```python
movies.isnull().any().any()
```

```python
ratings.shape
```

```python
tags.shape
```

```python
tags.isnull().any().any()
```

```python
tags=tags.dropna()
```

dropna() removes missing values.



By default, it removes rows with missing values.



Assigning the result back to tags updates the variable.



It does not remove empty strings ('') by default.

```python
movies.isnull().sum()
```

```python
movies.isnull().sum().sum()
```

True means at least one missing value exists; False means none were detected. Empty strings ('') are not considered missing by default.
